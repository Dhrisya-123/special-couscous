```python
from atom import Atom
filename = "6B1E.pdb" 
def read_pdb(filename):
    coords = []
    
    with open(filename) as f:
        for line in f:
            line=line.lstrip().rstrip()
            
            if not line:
                continue
            if line.startswith("ATOM") or line.startswith("HETATM"):
                atom=Atom()
                atom.set_data(line)
                coords.append(atom)
                
    return coords
```

```python
n=read_pdb(filename)
print (n)
```

```python
file = read_pdb(filename)  
dist = []
m=file[1]
for atom in file:
    val=atom.distance(m)
    dist.append(val)
print (dist)
```
